## 4.1 Hvorfor bruke løkker?

Datamaskiner er spesielt gode på én ting mennesker sliter med: å gjøre den **samme oppgaven igjen og igjen**, veldig raskt, uten å bli lei. Dette gjør vi med **løkker**.

I Python har vi to typer løkker:
- **while-løkke**: gjentas *så lenge* en betingelse er sann
- **for-løkke**: gjentas et *forhåndsbestemt* antall ganger (f.eks. for hvert element i en liste)

## 4.2 while-løkker

En `while`-løkke kjører koden inni seg helt til betingelsen blir `False`.

```python
while betingelse:
    # kode som gjentas
```

**NB:** Du må selv sørge for at betingelsen på et tidspunkt blir usann, ellers får du en **evig løkke** som aldri stopper!

````python
x = 1

while x <= 5:
    print("x er nå", x)
    x = x + 1   # Øker x med 1 for hver runde - dette er avgjørende!

print("Løkken er ferdig")
````

### 💡 Eksempel: Nedtelling


````python
teller = 5

while teller > 0:
    print(teller)
    teller = teller - 1

print("Tiden er ute!")
````

## 4.3 for-løkker

En `for`-løkke går gjennom hvert element i en liste (eller et listeliknende objekt), og kjører koden i løkken én gang per element.

```python
for element in liste:
    # kode som kjøres for hvert element
```

````python
fag = ["Programmering", "Norsk", "Matte", "Engelsk"]

for f in fag:
    print("Du har faget:", f)
````

### 4.3.1 range()-funksjonen

Ofte vil vi bare kjøre en løkke et gitt antall ganger, uten å ha en liste ferdig. Da bruker vi `range()`.

| Bruk | Resultat (liknende liste) |
|---|---|
| `range(n)` | `[0, 1, 2, ..., n-1]` |
| `range(start, stopp)` | `[start, start+1, ..., stopp-1]` |
| `range(start, stopp, skritt)` | `[start, start+skritt, ...]` opp til (men ikke med) `stopp` |

````python
for i in range(5):
    print(i)
````

````python
for i in range(2, 10):
    print(i)
````

````python
for i in range(0, 20, 5):   # start=0, stopp=20, skritt=5
    print(i)
````

### 💡 Eksempel: Summere tall i en løkke

````python
sum_tall = 0

for tall in range(1, 11):   # 1 til og med 10
    sum_tall = sum_tall + tall

print("Summen av tallene 1 til 10 er:", sum_tall)
````
